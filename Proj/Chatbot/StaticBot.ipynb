# Static Chatbot

This creates a static chatbot - one that is not interactive in the usual sense. To communicate with it, you have to edit your question, which will be fed to the chatbot.

## Requirements

You will need a/an:
  - Environment : You can create one using the built-in Python venv, or use conda or micromamba.
  - [Groq](https://groq.com/) API Key : which you can obtain from [here](https://console.groq.com/keys) for free. 
  - .env file: Then copy the generated key and paste it in the .env_example file then rename the file from .env_example to .env 

You are now ready to roll!

```python
# ! pip install dotenv langchain langchain-groq # Install these libraries if necessary
from langchain_groq import ChatGroq
from langchain_core.prompts import PromptTemplate
from dotenv import load_dotenv
load_dotenv()
import os

KEY = os.environ.get("GROQ_API_KEY")
```

## The Interesting part

If you have a question, just edit the sentence *How did von Neumann revolutionize Computer Science* and dont touch the quotation marks " " .

My favorite part is seeing how the AI **"Thinks"**, and the following should show you that part.

```python
llm = ChatGroq(api_key=KEY,
               model='qwen/qwen3-32b', 
               temperature=0) # This is to get deteministic & factual response

question = " How did von Neumann revolutionize Computer Science "
response = llm.invoke(question)
print(response.content)
```
